<h1 style="text-align: center;">The Kernel (The Brain 🧠)</h1>

<font size="4">In the top right corner of your screen, you should see "Python 3". This is the Kernel.</font>
<font size="4">The Kernel is the computational engine that executes your code.</font>

- <font size="4">Circle is Empty (⚪): The kernel is idle and ready.</font>

- <font size="4">Circle is Full (⚫): The kernel is busy calculating (e.g., loading a huge dataset). You must wait.</font>

<font size="4">**Troubleshooting**:</font>

<font size="4">If your notebook gets stuck or stops running code, you can reset the Kernel by clicking the Restart button (the circular arrow ↻) in the toolbar at the top. This clears all memory variables but keeps your code written down.</font>

<font size="4">Just to show you why we use Python, run the cell below. It uses matplotlib (a plotting library) to generate some data and plot it.</font>

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate some fake data (e.g., Reaction Times for 100 participants)
# Mean = 500ms, SD = 50ms
reaction_times = np.random.normal(loc=500, scale=50, size=100)

# 2. Create a simple histogram
plt.figure(figsize=(8, 5))
plt.hist(reaction_times, bins=15, color='skyblue', edgecolor='black')
plt.title('Distribution of Reaction Times')
plt.xlabel('Time (ms)')
plt.ylabel('Number of Participants')

# 3. Show the plot
plt.show()
```
