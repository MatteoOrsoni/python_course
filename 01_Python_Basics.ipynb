<h1 style="text-align: center;">Naming Conventions</h1>

<font size="4">Python is case-sensitive: `Age`, `age`, and `AGE` are three completely different variables.

<font size="4">To keep code readable and consistent, Python researchers follow the **PEP 8** style guide. The most important rule for variables is to use snake_case.

<font size="4">**The Rules of Snake Case:**
- <font size="4">Use all lowercase letters.
- <font size="4">Separate words with underscores _.

<font size="4">Be descriptive (avoid single letters like x or a unless it's for math).

| <font size="4">Status | <font size="4">Variable Name  | <font size="4">Why?                                                   |
| ------ | -------------- | ------------------------------------------------------ |
| ✅ <font size="4">Good | <font size="4">reaction_time  | <font size="4">Clear, readable, standard Python style.                |
| ✅ <font size="4">Good | <font size="4">participant_id | <font size="4">Descriptive.                                           |
| ❌ <font size="4">Bad  | <font size="4">ReactionTime   | <font size="4">This is "CamelCase" (used in Java/Matlab, not Python). |
| ❌ <font size="4">Bad  | <font size="4">reaction time  | <font size="4">ERROR: You cannot have spaces in variable names!       |
| ❌ <font size="4">Bad  | <font size="4">rt             | <font size="4">Too vague.        |

<font size="4">Variable names must follow strict rules to avoid crashing your code.

<h1 style="text-align: center;">Doing Math with Variables</h1>

<font size="4">You can use variables to perform calculations just like a calculator. This is the foundation of data analysis (e.g., calculating accuracy, converting units, or standardizing scores).

| <font size="4">Symbol | <font size="4">Operation        | <font size="4">Example | <font size="4">Result              |
| ------ | ---------------- | ------- | ------------------- |
| <font size="4">+      | <font size="4">Addition         | <font size="4">10 + 5  | <font size="4">15                  |
| <font size="4">-      | <font size="4">Subtraction      | <font size="4">10 - 5  | <font size="4">5                   |
| <font size="4">*      | <font size="4">Multiplication   | <font size="4">10 * 5  | <font size="4">50                  |
| <font size="4">/      | <font size="4">Division         | <font size="4">10 / 2  | <font size="4">5.0 (Always float!) |
| <font size="4">**     | <font size="4">Exponent (Power) | <font size="4">2 ** 3  | <font size="4">8 ($2^3$)           

<font size="4">You can, for example, convert Reaction Time (RT) from milliseconds to seconds.

```python
rt_ms = 450         # RT in milliseconds
rt_sec = rt_ms / 1000   # Divide by 1000

print("RT in ms:", rt_ms)
print("RT in sec:", rt_sec)
```

<font size="4">Notice that division / always returns a Float (decimal), even if the result is a whole number (10 / 2 = 5.0).
If you need an Integer result (e.g., for counting groups), use Floor Division `//`, which drops the decimal part.

```python
# Standard Division
print(5 / 1.7)   # Output: 2.5

# Floor Division (Integer)
print(5 // 1.7)  # Output: 2 (The .5 is thrown away)
```

<h1 style="text-align: center;">Working with Text (Strings)</h1>

<font size="4">**The Magic of f-strings**

<font size="4">Often you need to insert a variable inside a sentence (e.g., printing a result or creating a filename). 
The modern way to do this is using f-strings (formatted strings).

- <font size="4">Put a little f before the opening quote.
- <font size="4">Put variables inside curly brackets {}.

```python
sub_id = 42
condition = "stress"

# The f-string way:
print(f"Subject {sub_id} was in the {condition} group.")

# You can even do filenames:
filename = f"data_subject_{sub_id}.csv"
print(filename)
```

<font size="4">**Cleaning Text Methods**
<font size="4">Participants are messy typists. They might type " male " instead of "male". Python strings have built-in "methods" to fix this.

| <font size="4">Method   | <font size="4">Action                  | <font size="4">Example Input | <font size="4">Result  |
| -------- | ----------------------- | ------------- | ------- |
| <font size="4">.upper() | <font size="4">Convert to ALL CAPS     | <font size="4">"hello"       | <font size="4">"HELLO" |
| <font size="4">.lower() | <font size="4">Convert to lowercase    | <font size="4">"Male"        | <font size="4">"male"  |
| <font size="4">.strip() | <font size="4">Remove start/end spaces | <font size="4">"  data "     | <font size="4">"data"  |

```python
raw_input = "  FeMaLe  "

# Chain methods together: Strip spaces THEN make lowercase
clean_input = raw_input.strip().lower()

print(f"Original: '{raw_input}'")
print(f"Cleaned:  '{clean_input}'")
```

<h1 style="text-align: center;">Lists: Storing Sequences</h1>

<font size="4">In research, we rarely work with single numbers. We work with _sequences_ of data: a list of reaction times, a list of condition names, or a list of participant ages.

<font size="4">In Python, a List is a collection of items wrapped in square brackets `[]`.

```python
# A list of Integers (e.g., ages)
ages = [21, 25, 19, 30]

# A list of Strings (e.g., stimuli)
stimuli = ["Face", "House", "Shoe", "Chair"]

# A mixed list (Valid, but rare in analysis)
mixed = [101, "Control", 0.45]

print(ages)
print(stimuli)
```

<font size="4">**Indexing**</font>

<font size="4">Lists are ordered. To grab a specific item, you use its Index (position).

<font size="4">⚠️ **Crucial Rule**: Python counts from **ZERO**!

<font size="4">1st item = Index 0

<font size="4">2nd item = Index 1

```python
# Get the first item (Index 0)
first_stim = stimuli[0] 

# Get the third item (Index 2)
third_stim = stimuli[2]

# Get the LAST item (Special shortcut: -1)
last_stim = stimuli[-1]

print(f"First: {first_stim}, Third: {third_stim}, Last: {last_stim}")
```

<font size="4">**List Methods: Adding things**
    
<font size="4">Lists are mutable (changeable). You can add new participants as they arrive.

<font size="4">`.append(item)`: Adds an item to the end of the list.

<font size="4">To insert an item at a specific position (instead of just the end), you use the `.insert()` method.

<font size="4">It takes two arguments: `.insert(index, item)`.

```python
rt_list = [450, 500, 480]

# New trial comes in:
rt_list.append(600)

print(rt_list) 
# Output: [450, 500, 480, 600]

stimuli = ["Face", "House", "Shoe"]

# I want "Tree" to be the 2nd item (Index 1)
stimuli.insert(1, "Tree")

print(stimuli)
# Output: ["Face", "Tree", "House", "Shoe"]
```

<font size="4">**Dictionary of Lists (Column-based), and Lists of Dictionaries (Row-based)**</font> 

| <font size="4">Structure     | <font size="4">Analogy              | <font size="4">Access Pattern                         |
| ------------- | -------------------- | -------------------------------------- |
| <font size="4">List of Dicts | <font size="4">Rows of participants | <font size="4">data[0]['age'] <font size="4">(Row $\rightarrow$ Col) |
| <font size="4">Dict of Lists | <font size="4">Columns of variables | <font size="4">data['age'][0] <font size="4">(Col $\rightarrow$ Row) |

<font size="4">Imagine a spreadsheet where you have one column for "IDs" and one column for "Scores".

<font size="4">In Python, this is a Dictionary where every value is a List.

```python
#####################################################
################### DICT OF LISTS ###################
#####################################################

dataset = {
    'ids':    [1, 2, 3],
    'scores': [85, 90, 78],
    'groups': ['Control', 'Exp', 'Control']
}

# First you pick the Column ('key'), THEN the Row [index].

# Get all scores
all_scores = dataset['scores'] # [85, 90, 78]

# Get the score of the first participant (Index 0)
first_score = dataset['scores'][0]

print(f"All Scores: {all_scores}")
print(f"First Score: {first_score}")
```

```python
#####################################################
################### LIST OF DICTS ###################
#####################################################

data_log = [
    {'id': 1, 'score': 85, 'group': 'Control'},
    {'id': 2, 'score': 90, 'group': 'Exp'},
    {'id': 3, 'score': 78, 'group': 'Control'}
]

# First index [ ] selects the Row (Dictionary)
# Second index [' '] selects the Column (Key)

# 1. Get the second participant (Index 1)
p2 = data_log[1]
print(p2)  # {'id': 2, 'score': 90, 'group': 'Exp'}

# 2. Get the score of the second participant directly
p2_score = data_log[1]['score']
print(f"P2 Score: {p2_score}")
```

<h1 style="text-align: center;">Control Flow: Logic and Loops</h1>

<font size="4">**Control Flow**: making decisions (`if`) and repeating actions (`for` loops).

<font size="4">This is essential for:

- <font size="4">**Analysis**: "If RT < 100ms, mark as outlier."
- <font size="4">**Automation**: "For every participant file in the folder, load it."

<font size="4">**The if Statement (Logic)**

<font size="4">You can run code only if a condition is true.

- <font size="4">`==` (Equals)
- <font size="4">`!=` (Not equals)
- <font size="4">`>` (Greater than)
- <font size="4">`<` (Less than)
<font size="4">**Important**: Python uses indentation (4 spaces or Tab) to know what code belongs inside the if block.

```python
raw_input = input("Please insert an integer between 10 and 3000: ")
rt = int(raw_input)

if rt < 100:
    print("⚠️ Warning: Too fast! Possible false start.")
elif rt > 1000:
    print("⚠️ Warning: Too slow!")
else:
    print("✅ Valid trial.")
```

<font size="4">**The for Loop (Iteration)**

<font size="4">If you have a list of 100 reaction times, you don't want to write 100 if statements. You use a loop.

<font size="4">A for loop takes each item from a list, one by one, and puts it in a temporary variable (e.g., item).

```python
# A list of RTs
rt_data = [450, 90, 600, 1200, 500]

print("Starting analysis...")

for val in rt_data:
    # This code runs 5 times (once for each number)
    if val < 100:
        print(f"Value {val} is too fast!")
    elif val > 1000:
        print(f"Value {val} is too slow!")
    else:
        print(f"Value {val} is okay.")

print("Analysis complete.")
```

<font size="4">**The `range()` function**

<font size="4">Sometimes you just want to repeat something N times (e.g., "Run 10 trials").

```python
# Repeat 3 times
for i in range(3):
    print(f"Trial {i+1} starting...")
```
